# Libraries and Data

```python
# Import libraries
import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Set figure size to 10,6 and sns style to whitegrid
plt.figure(figsize=(10,6))
sns.set(style='whitegrid')
```

```python
# Load data and see first lines
df = pd.read_csv('../data/demand_forecasting.csv')
df.head()
```

# Explore Dataset

```python
# Check which values besides zeros column Epidemic has
df["Epidemic"].value_counts()
```

```python
# Check datatypes
df.dtypes
```

```python
# Change datatype of column date to datetime
df['Date'] = pd.to_datetime(df['Date'])
df.dtypes
```

```python
# Get info of dataframe
df.info()
```

```python
# Check of there are any na values
df.isna().sum().sum()
```

```python
# Check of there are duplicates
df.duplicated().sum()
```

```python
# Get statistical information on numeric data
df.describe().T
```

```python
# Get statistical information on categorical data
df.describe(include='object').T
```

# Feature Engineering and Analysis

```python
# Create year, month, day and weekday columns
df['Year'] = df['Date'].dt.year
df['Month'] = df['Date'].dt.month
df['Day'] = df['Date'].dt.day
df['Weekday'] = df['Date'].dt.day_name()
df.sample(3)
```

```python
# Create column of price less discount
df['Discounted Price'] = df['Price'] * (1 - df['Discount'] / 100)
df.head()
```

```python
# Calculate stock utilization
df['Sell Through Rate'] = df['Units Sold'] / df['Inventory Level']
df.head(1)
```

```python
# Statistics of Sell Through Rate
df['Sell Through Rate'].describe()
```

```python
# Calculate mean, sum and std of demand by category, sort by sum descending, make it a dataframe
df.groupby('Category')['Demand'].agg(['mean', 'sum', 'std']).sort_values(by='sum', ascending=False).reset_index()
```

```python
# Calculate mean demand by region and seasonality
df.groupby(['Region', 'Seasonality'])['Demand'].mean()
```

```python
# See promotion impact
df.groupby('Promotion')['Demand'].mean()
```

```python
# Monthly average deamand by category (pivot table)
pd.pivot_table(df, values='Demand', index='Month', columns='Category', aggfunc='mean')
```

# Visualizations

```python
# Demand distribution visualization
sns.histplot(df['Demand'], bins=20, kde=True)
plt.title('Demand Distribution')
plt.show()
```

```python
# Scatter plot of units solld and inventory
sns.scatterplot(data = df, x='Inventory Level', y='Units Sold')
plt.title('Inventory vs Units Sold')
plt.show()
```

```python
# Boxplot demand by category
sns.boxplot(data=df, x='Category', y='Demand', hue='Category')
plt.xticks(rotation=45)
plt.title('Demand by Category')
plt.show()
```

```python
# Boxplot demand by weather condition
sns.boxplot(data=df, x='Weather Condition', 
            y='Demand', 
            hue="Weather Condition")
plt.xticks(rotation=45)
plt.title('Demand by Weather Condition')
plt.show()
```

# Time Series Analysis

```python
# Create variable for Average Monthly Demand
monthly_demand = df.groupby('Month')['Demand'].mean()
monthly_demand
```

```python
# Show bar plot for Monthly Demand
monthly_demand.plot(kind = 'bar')
plt.title('Average Demand by Month')
plt.xticks(rotation=0)
plt.show()
```

```python
# Plot with sns
monthly_demand = pd.DataFrame(monthly_demand)
sns.barplot(data=monthly_demand, x='Month', y='Demand', hue='Month', legend=False, palette='dark:green')
plt.show()
```

```python
# Create variable for overall daily trend
daily_demand = df.groupby('Date')['Demand'].sum()
daily_demand
```

```python
# Visualize daily trend
daily_demand.plot()
plt.title('Total Daily Deamand Over Time')
plt.xlabel('Date')
plt.ylabel('Demand')
plt.show()
```

```python
# Plot with sns
daily_demand = pd.DataFrame(daily_demand)
sns.lineplot(data=daily_demand, x='Date', y='Demand')
plt.xticks(rotation=45)
plt.show()
```

```python
# Show Promotion vs. Demand
sns.barplot(data=df, x='Promotion', y='Demand', hue='Promotion', legend=False)
plt.title('Promotion Impact on Demand')
plt.show()
```

```python
# Show Discounted Price vs Demand (Scatter)
sns.scatterplot(data=df, x='Discounted Price', y='Demand')
plt.title('Discounted price vs Demand')
plt.show()
```

```python
# Demand by seasonality
df.groupby('Seasonality')['Demand'].mean().plot(kind='bar', title='Demand by Seasonality')
plt.show()
```

```python
# Plot with sns
seasonality = pd.DataFrame(df.groupby('Seasonality')['Demand'].mean())
sns.barplot(data=seasonality, x='Seasonality', y='Demand', hue='Seasonality', legend=False)
plt.plot()
```

```python
# Epidemic impact on Demand
df.groupby('Epidemic')['Demand'].mean().plot(kind='bar', title='Demand by Epidemic')
plt.show()
```

```python
# Plot with sns
epidemic = pd.DataFrame(df.groupby('Epidemic')['Demand'].mean())
sns.barplot(data=epidemic, x='Epidemic', y='Demand', hue='Epidemic', legend=False)
plt.show()
```
